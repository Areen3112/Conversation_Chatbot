### MESSAGE HISTORY

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}

def get_session_history(session_id: str) -> BaseChatMessageHistory:
    # Implement your logic to retrieve the chat history for the given session_id
    # This could involve querying a database, reading from a file, etc.
    # For demonstration purposes, we'll return an empty history
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]


with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable": {"session_id": "chat1"}}

```

```python
response = with_message_history.invoke(
    [
        HumanMessage(content="Hi , I am Areen Joshi an AI engineering student. ")

    ],
    config=config
)
```

```python
with_message_history.invoke(
    [
        HumanMessage(content = "What is my name ?")
    ],
    config = config
)
```

### PROMPT TEMPLATES


```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant that remembers the user's name."),
        MessagesPlaceholder(variable_name="messages"),
    ]
)


chain = prompt | model
```

```python
chain.invoke(
    {
        "messages":[HumanMessage(content = "Hi My name is Areen Joshi. ")]
    }
)
```

```python
with_message_history = RunnableWithMessageHistory(chain, get_session_history)
```

```python
config = {"configurable": {"session_id": "chat3"} }
response = with_message_history.invoke(
    [HumanMessage(content = "Hi My name is Areen Joshi. ")],
    config = config
)

response 
```

```python
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant that remembers the user's name. Answer  questions to the best of your abilityin {language}."),
        MessagesPlaceholder(variable_name="messages"),
    ]
)


chain = prompt | model
```

```python
reponse= chain.invoke(
    {
        "messages":[HumanMessage(content = "Hi My name is Areen Joshi. ")],
        "language": "Hindi"
    }
)

response.content
```

```python
with_message_history= RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages = "messages"
)
```

### MANAGING CONVERSATIONAL HISTORY

```python
from langchain_core.messages import SystemMessage, trim_messages
trimmer = trim_messages(
    max_tokens = 70,
    strategy= 'last', 
    token_counter= model,
    include_system=True, 
    allow_partial=False, 
    start_on= "human"
)


messages = [
    SystemMessage(content="you're a good assistant"),
    HumanMessage(content="hi! I'm bob"),
    AIMessage(content="hi!"),
    HumanMessage(content="I like vanilla ice cream"),
    AIMessage(content="nice"),
    HumanMessage(content="whats 2 + 2"),
    AIMessage(content="4"),
    HumanMessage(content="thanks"),
    AIMessage(content="no problem!"),
    HumanMessage(content="having fun?"),
    AIMessage(content="yes!"),
]

trimmer.invoke(messages)
```
